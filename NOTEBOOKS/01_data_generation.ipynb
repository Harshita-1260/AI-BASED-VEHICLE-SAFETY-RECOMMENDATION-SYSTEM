```python
import numpy as np
np.random.seed(42)
import pandas as pd 
print("libraries imported success")
```

```python
n = 5000

# -----------------------------
# DRIVER FEATURES
# -----------------------------

driver_age = np.random.randint(22, 60, n)

driving_hours = np.random.uniform(1, 12, n).round(1)

speed = np.random.randint(35, 110, n)

harsh_braking = np.random.poisson(2, n)

harsh_acceleration = np.random.poisson(2, n)

# Drowsiness increases with driving hours
drowsiness_score = np.clip(
    (driving_hours * 8) + np.random.normal(20, 10, n),
    0,
    100
).round(1)


# -----------------------------
# VEHICLE FEATURES
# -----------------------------

engine_temperature = np.random.normal(92, 8, n).round(1)

engine_vibration = np.random.normal(3.5, 1, n).round(2)

battery_voltage = np.random.normal(12.6, 0.5, n).round(2)

oil_pressure = np.random.normal(42, 8, n).round(1)

tyre_pressure = np.random.normal(33, 3, n).round(1)

vehicle_age = np.random.randint(1, 16, n)

mileage = np.random.randint(50000, 900000, n)

previous_breakdowns = np.random.poisson(1, n)

previous_accidents = np.random.poisson(0.5, n)


# -----------------------------
# ENVIRONMENT FEATURES
# -----------------------------

weather = np.random.choice(
    ["Clear", "Rain", "Fog"],
    n,
    p=[0.60, 0.25, 0.15]
)

road_type = np.random.choice(
    ["Highway", "City", "Mountain"],
    n,
    p=[0.50, 0.30, 0.20]
)


# -----------------------------
# CREATE DATAFRAME
# -----------------------------

df = pd.DataFrame({
    "Driver_Age": driver_age,
    "Driving_Hours": driving_hours,
    "Speed": speed,
    "Harsh_Braking": harsh_braking,
    "Harsh_Acceleration": harsh_acceleration,
    "Drowsiness_Score": drowsiness_score,
    "Engine_Temperature": engine_temperature,
    "Engine_Vibration": engine_vibration,
    "Battery_Voltage": battery_voltage,
    "Oil_Pressure": oil_pressure,
    "Tyre_Pressure": tyre_pressure,
    "Vehicle_Age": vehicle_age,
    "Mileage": mileage,
    "Previous_Breakdowns": previous_breakdowns,
    "Previous_Accidents": previous_accidents,
    "Weather": weather,
    "Road_Type": road_type
})

print("Dataset Generated Successfully ✅")
print("Dataset Shape:", df.shape)

df.head()
```

```python
## AI Target columns
df.shape
```

```python
# ==========================================
# DRIVER SAFETY RISK SCORE
# ==========================================

driver_score = (
    (df["Driving_Hours"] > 8).astype(int) * 2 +
    (df["Speed"] > 85).astype(int) * 2 +
    (df["Harsh_Braking"] > 3).astype(int) +
    (df["Harsh_Acceleration"] > 3).astype(int) +
    (df["Drowsiness_Score"] > 70).astype(int) * 2 +
    (df["Previous_Accidents"] > 1).astype(int)
)


# ==========================================
# VEHICLE BREAKDOWN RISK SCORE
# ==========================================

breakdown_score = (
    (df["Engine_Temperature"] > 100).astype(int) * 2 +
    (df["Engine_Vibration"] > 5).astype(int) * 2 +
    (df["Battery_Voltage"] < 12).astype(int) +
    (df["Oil_Pressure"] < 30).astype(int) +
    (df["Tyre_Pressure"] < 28).astype(int) +
    (df["Vehicle_Age"] > 10).astype(int) +
    (df["Mileage"] > 600000).astype(int) +
    (df["Previous_Breakdowns"] > 2).astype(int)
)


# ==========================================
# FINAL TARGETS
# ==========================================

df["Driver_Risk"] = np.where(
    driver_score >= 4,
    1,
    0
)

df["Breakdown_Risk"] = np.where(
    breakdown_score >= 4,
    1,
    0
)


print("Target Columns Created Successfully ✅")

print("\nDriver Risk:")
print(df["Driver_Risk"].value_counts())

print("\nBreakdown Risk:")
print(df["Breakdown_Risk"].value_counts())
```
