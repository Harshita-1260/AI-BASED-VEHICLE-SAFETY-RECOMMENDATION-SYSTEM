```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/processed/truck_processed.csv")
df.head()
```

```python
x=df.drop(columns=["Driver_Risk","Breakdown_Risk"])
y=df["Driver_Risk"]
```

```python
numerical_features = [
    "Driver_Age",
    "Driving_Hours",
    "Speed",
    "Harsh_Braking",
    "Harsh_Acceleration",
    "Drowsiness_Score",
    "Previous_Accidents"
]

categorical_features = [
    "Weather",
    "Road_Type"
]

print("Numerical Features:", numerical_features)
print("Categorical Features:", categorical_features)
```

```python

from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42,stratify=y)   
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer

numerical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer([
    ("num", numerical_pipeline, numerical_features),
    ("cat", categorical_pipeline, categorical_features)
])

print("Preprocessor created successfully")
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline

driver_model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", RandomForestClassifier(
        n_estimators=200,
        random_state=42,
        class_weight="balanced"
    ))
])
```

```python
driver_model.fit(x_train, y_train)
```

```python
y_pred = driver_model.predict(x_test)
y_prob = driver_model.predict_proba(x_test)[:, 1]
```

```python
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score

print("Classification Report:")
print(classification_report(y_test, y_pred))

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))

print("Accuracy score:")
print("accuracy:", driver_model.score(x_test, y_test))
```

```python
print("accuracy:", driver_model.score(x_test, y_test))
```

```python
# Get feature names after preprocessing
feature_names = driver_model.named_steps["preprocessor"].get_feature_names_out()

# Get feature importance from Random Forest
importances = driver_model.named_steps["classifier"].feature_importances_

# Create DataFrame
feature_importance_df = pd.DataFrame({
    "Feature": feature_names,
    "Importance": importances
})

# Sort from highest to lowest
feature_importance_df = feature_importance_df.sort_values(
    by="Importance",
    ascending=False
)

# Display top 10 features
print("Top 10 Important Features:")
display(feature_importance_df.head(10))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(10, 6))

top_features = feature_importance_df.head(10)

sns.barplot(
    data=top_features,
    x="Importance",
    y="Feature"
)

plt.title("Top 10 Features - Driver Safety Prediction")
plt.xlabel("Feature Importance")
plt.ylabel("Feature")

plt.tight_layout()

plt.savefig(
    "../REPORTS/FIGURES/driver_safety_feature_importance.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
